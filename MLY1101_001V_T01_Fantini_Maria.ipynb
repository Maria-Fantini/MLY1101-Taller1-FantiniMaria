# Fuente 1 — API 1

**Nombre de la API:** AnimeChan

**Documentación:**  Animechan es un servicio API de citas e información sobre anime que proporciona a los desarrolladores acceso a una vasta colección de contenido de anime. La API ofrece el número de episodios, resúmenes detallados de las series, información sobre los personajes y citas memorables de miles de series de anime. (https://animechan.io/docs)

**Endpoint utilizado:**  https://api.animechan.io/v1/anime

**Descripción de los datos:**  El dataset posee: El ID, el nombre en ingles, el nombre original (alternativo / japones), el ID de MyAnimeList, la cantidad de episodios y la cantidad de escenas/capitulos que hay dentro del propio capítulo (esto es para quienes al leer el manga/manwha sepan que la cantidad de capítulos que está conteniendo la adaptación animada)

**Relación con el objetivo:**  Esta primera API nos ayuda a saber en este caso que información contiene sobre cada anime de forma general, aunque no revela información sobre la trama de cada serie.


````python
# CONFIGURACIÓN API 1
API1_URL = 'https://api.animechan.io/v1/anime'
API1_MIN_REGISTROS = 200

# Modifique headers o params si la API lo requiere
api1_headers = {}
api1_params = {}
````

### Paginación API 1

Si la API utiliza paginación, reemplace el bloque anterior por un ciclo que continúe hasta obtener al menos 200 registros o hasta que no existan más páginas.


````python
# EJEMPLO BASE DE PAGINACIÓN — API 1
# Ajustar nombres de parámetros y estructura de respuesta según la API.

api1_registros = []
page = 1

while len(api1_registros) < API1_MIN_REGISTROS:
     params = {'page': page}
     response = requests.get(API1_URL, params=params, timeout=30)
     response.raise_for_status()
     data = response.json()

     nuevos = data['data']  # Ajustar
     if not nuevos:
         break

     api1_registros.extend(nuevos)
     page += 1

print('Registros API 1:', len(api1_registros))
````

````python
# GUARDAR DATASET API 1
API1_ARCHIVO = OUTPUT_DIR / 'dataset_api_1_animechan.json'

with open(API1_ARCHIVO, 'w', encoding='utf-8') as f:
    json.dump(api1_registros, f, ensure_ascii=False, indent=2)

print('Archivo generado:', API1_ARCHIVO)
print('Registros guardados:', len(api1_registros))
````

# Fuente 2 — API 2

**Nombre de la API:** poetrydb

**Documentación:**  https://openlibrary.org/developers/api

**Endpoint utilizado:**  https://poetrydb.org/random/200

**Descripción de los datos:**  La API PoetryDB proporciona información relacionada con poemas. Para cada registro se obtienen principalmente los siguientes datos: título del poema (title), autor (author), contenido del poema dividido en líneas (lines) y cantidad de líneas (linecount). Mediante el endpoint utilizado se solicitaron 200 poemas aleatorios en una sola consulta.

**Relación con el objetivo:**  A diferencia de la API 1, PoetryDB permite trabajar directamente con información literaria y textual. Los datos obtenidos permiten analizar características de distintos poemas, como sus autores, títulos y extensión en cantidad de líneas. Además, el contenido almacenado en lines puede ser utilizado posteriormente para realizar análisis del texto y obtener información adicional a partir de los poemas. Cabe detallar que no lleva paginación


````python
# CONFIGURACIÓN API 2
import requests
import time

API2_URL = 'https://poetrydb.org/random/200'
API2_MIN_REGISTROS = 200

api2_registros = []
````

````python
# CONSUMO API 2
api2_registros = []

response = requests.get(API2_URL, timeout=30)
response.raise_for_status()

data = response.json()

nuevos = data

api2_registros.extend(nuevos)

print('Registros API 2:', len(api2_registros))
````

````python
# GUARDAR DATASET API 2
import json

API2_ARCHIVO = OUTPUT_DIR / 'dataset_api_2.json'

with open(API2_ARCHIVO, 'w', encoding='utf-8') as f:
    json.dump(api2_registros, f, ensure_ascii=False, indent=2)

print('Archivo generado:', API2_ARCHIVO)
print('Registros guardados:', len(api2_registros))
````
